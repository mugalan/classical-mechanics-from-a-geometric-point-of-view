# Thermalization of a 2-particle system

We consider particles that move freely between instantaneous collisions. Following the idealization used here, two colliding particles coincide at the collision instant. The discussion below establishes the consequences of a single elastic collision and then uses statistical assumptions to motivate thermal equilibrium.

Let two particles have masses $m_1,m_2$, with position representations $x_1,x_2$ in an inertial frame $\mathbf e$. Let $\mathbf b$ be the center of mass frame, with axes parallel to those of $\mathbf e$. If $o$ represents the center of mass in $\mathbf e$, then
$$
x_i=o+X_i,\qquad
m_1X_1+m_2X_2=0,\qquad
m_1\dot X_1+m_2\dot X_2=0.
$$
Consequently, the total momentum and kinetic energy are
$$
m_1\dot x_1+m_2\dot x_2=(m_1+m_2)\dot o,
$$
$$
\mathrm{KE}
=\frac{m_1+m_2}{2}\|\dot o\|^2
+\frac{m_1}{2}\|\dot X_1\|^2
+\frac{m_2}{2}\|\dot X_2\|^2.
$$

Use a prime to denote quantities immediately after collision. In the absence of an external impulse, conservation of momentum gives
$$
\dot o'=\dot o.
$$
Thus, **the collision leaves the center of mass velocity unchanged**.

An **elastic collision** also conserves total kinetic energy. Since $\dot o'=\dot o$, it follows that
$$
\frac{m_1}{2}\|\dot X_1\|^2+\frac{m_2}{2}\|\dot X_2\|^2
=
\frac{m_1}{2}\|\dot X_1'\|^2+\frac{m_2}{2}\|\dot X_2'\|^2.
$$
Using
$$
\dot X_2=-\frac{m_1}{m_2}\dot X_1,\qquad
\dot X_2'=-\frac{m_1}{m_2}\dot X_1',
$$
we obtain
$$
\|\dot X_1'\|=\|\dot X_1\|,\qquad
\|\dot X_2'\|=\|\dot X_2\|.
$$
Hence, **each particle’s speed in the center of mass frame is unchanged; only its direction can change**. Under our collision idealization, the particles move along a common straight line through the origin before collision and along a possibly different such line afterwards.

To examine energy redistribution through repeated collisions, note the identity
$$
\begin{aligned}
\dot o\cdot(\dot X_2-\dot X_1)
&=\dot o\cdot(\dot x_2-\dot x_1)\\
&=\frac{
m_2\|\dot x_2\|^2-m_1\|\dot x_1\|^2
+(m_1-m_2)\dot x_1\cdot\dot x_2
}{m_1+m_2}.
\end{aligned}
$$

Now consider two particles in a gas confined within a stationary container, with elastic collisions between particles and with the walls. Let angle brackets denote long-time averages in the container’s rest frame. We adopt two statistical idealizations:
$$
\left\langle\dot o\cdot(\dot x_2-\dot x_1)\right\rangle=0,
\qquad
\left\langle\dot x_1\cdot\dot x_2\right\rangle=0.
$$
The first assumes no persistent alignment between the center of mass velocity and the relative velocity. The second is motivated by approximately independent particle velocities with zero mean. **Neither follows from the collision laws or the absence of a preferred direction alone.**

Averaging the identity above then gives
$$
\left\langle\frac{m_1}{2}\|\dot x_1\|^2\right\rangle
=
\left\langle\frac{m_2}{2}\|\dot x_2\|^2\right\rangle.
$$
Applying this argument pairwise motivates **equal mean translational kinetic energies at thermal equilibrium**, even for particles of different masses. Individual kinetic energies continue to fluctuate as collisions exchange energy. This is a heuristic argument, not a proof that repeated collisions necessarily establish equilibrium.

To summarize:

- Without a resultant external force, the center of mass velocity remains constant. Collisions with container walls can change the particles’ total momentum.
- Elastic collisions conserve total kinetic energy while redistributing it among particles; this conservation alone does not imply thermal equilibrium.
- Under the stated statistical assumptions, particles have equal mean translational kinetic energies at equilibrium.

# Question:

At thermal equilibrium, lighter particles have greater mean-square speeds. If they collide more frequently, does this mean that they approach equilibrium faster, or does the energy transferred per collision also matter?

##Answer:

Both matter. More frequent collisions provide more opportunities for energy exchange, but the approach to equilibrium also depends on how effectively those collisions redistribute energy.

At equilibrium,
$$
m_1\langle\|\dot x_1\|^2\rangle
=
m_2\langle\|\dot x_2\|^2\rangle,
$$
so lighter particles have greater root-mean-square speeds. This relation alone, however, determines neither their collision frequency nor how quickly equilibrium is reached.

Our discussion explains how elastic collisions redistribute energy and, under the stated statistical assumptions, motivates equal mean kinetic energies at equilibrium. It does not determine the time required to reach that state; doing so requires a description of both collision frequency and energy exchange per collision.

# SImulation example

This simulation illustrates energy redistribution through random elastic collisions between light and heavy particles in three dimensions.

Particles are initially assigned random velocities. At each collision step:

1. A pair is selected uniformly at random.
2. Its center of mass velocity remains unchanged.
3. The relative velocity is assigned a direction uniformly distributed over the sphere, while preserving its magnitude.

Each collision therefore conserves the pair’s total momentum and kinetic energy.

Repeated collisions allow energy to pass between particles. The simulation illustrates the tendency of the two mass groups to approach equal mean translational kinetic energies, measured in the rest frame of the gas:
$$
\left\langle \frac12 m_{\mathrm{light}}\|\dot x_{\mathrm{light}}\|^2\right\rangle
\approx
\left\langle \frac12 m_{\mathrm{heavy}}\|\dot x_{\mathrm{heavy}}\|^2\right\rangle.
$$
Here the averages are taken over particles of each type and smoothed over successive collisions. Individual energies and the finite-sample averages continue to fluctuate.

This provides a numerical illustration of the energy redistribution discussed in the notes. Random collision directions do not, by themselves, establish statistical independence, zero velocity correlations, or ergodicity.

Because pairs are selected uniformly without tracking their positions or speeds, the horizontal axis represents collision count, not physical time. The simulation therefore does not determine a physical thermalization timescale.

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Parameters and initialization
# ----------------------------
rng = np.random.default_rng(42)

N = 100
n_steps = 15000
window = 200
burn_in = 5000          # Chosen sampling cutoff; not proof of equilibration
sample_interval = 100
pairs_per_sample = 50

# Equal numbers of light and heavy particles
masses = np.concatenate([
    np.ones(N // 2),
    np.full(N - N // 2, 5.0),
])
rng.shuffle(masses)

light = masses == 1.0
heavy = masses == 5.0

velocities = 2.0 * rng.standard_normal((N, 3))

# Work in the center of mass frame of the whole gas
velocities -= np.average(velocities, axis=0, weights=masses)


def kinetic_energies():
    return 0.5 * masses * np.sum(velocities**2, axis=1)


def total_momentum():
    return np.sum(masses[:, None] * velocities, axis=0)


def random_unit_direction():
    """Uniformly distributed direction on the unit sphere."""
    phi = rng.uniform(0.0, 2.0 * np.pi)
    cos_theta = rng.uniform(-1.0, 1.0)
    sin_theta = np.sqrt(1.0 - cos_theta**2)

    return np.array([
        sin_theta * np.cos(phi),
        sin_theta * np.sin(phi),
        cos_theta,
    ])


def collide(i, j):
    """Conserve pair momentum and energy; randomize relative direction."""
    m1, m2 = masses[i], masses[j]
    v1 = velocities[i].copy()
    v2 = velocities[j].copy()

    v_com = (m1 * v1 + m2 * v2) / (m1 + m2)
    relative_speed = np.linalg.norm(v1 - v2)
    relative_new = relative_speed * random_unit_direction()

    velocities[i] = v_com + m2 / (m1 + m2) * relative_new
    velocities[j] = v_com - m1 / (m1 + m2) * relative_new


def mean_pair_dot():
    """Mean v_i · v_j over all distinct pairs, without an N × N array."""
    velocity_sum = velocities.sum(axis=0)
    numerator = (
        np.dot(velocity_sum, velocity_sum)
        - np.sum(velocities**2)
    )
    return numerator / (N * (N - 1))


def running_average(values):
    return np.convolve(
        values, np.ones(window) / window, mode="valid"
    )


initial_energy = kinetic_energies().sum()
initial_momentum = total_momentum().copy()

# Include the initial state at collision count zero
ke_light = np.empty(n_steps + 1)
ke_heavy = np.empty(n_steps + 1)
pair_dot_mean = np.empty(n_steps + 1)

ke_light_samples = []
ke_heavy_samples = []
pair_dot_samples = []

max_energy_error = 0.0
max_momentum_error = 0.0

# ----------------------------
# Simulation
# ----------------------------
for step in range(n_steps + 1):
    if step > 0:
        i, j = rng.choice(N, size=2, replace=False)
        collide(i, j)

    energies = kinetic_energies()

    ke_light[step] = energies[light].mean()
    ke_heavy[step] = energies[heavy].mean()
    pair_dot_mean[step] = mean_pair_dot()

    max_energy_error = max(
        max_energy_error,
        abs(energies.sum() - initial_energy),
    )
    max_momentum_error = max(
        max_momentum_error,
        np.linalg.norm(total_momentum() - initial_momentum),
    )

    if step >= burn_in and (step - burn_in) % sample_interval == 0:
        ke_light_samples.append(energies[light].copy())
        ke_heavy_samples.append(energies[heavy].copy())

        # Sample distinct particle pairs
        a = rng.integers(0, N, size=pairs_per_sample)
        b = rng.integers(0, N - 1, size=pairs_per_sample)
        b += b >= a

        pair_dot_samples.extend(
            np.sum(velocities[a] * velocities[b], axis=1)
        )

ke_light_all = np.concatenate(ke_light_samples)
ke_heavy_all = np.concatenate(ke_heavy_samples)
pair_dot_samples = np.asarray(pair_dot_samples)

steps = np.arange(n_steps + 1)
smooth_steps = steps[window - 1:]

print(
    "Maximum relative total-energy drift: "
    f"{max_energy_error / initial_energy:.3e}"
)
print(
    "Maximum total-momentum drift (simulation units): "
    f"{max_momentum_error:.3e}"
)
print(f"Post-cutoff mean KE, light: {ke_light_all.mean():.4f}")
print(f"Post-cutoff mean KE, heavy: {ke_heavy_all.mean():.4f}")

# ----------------------------
# Plot 1: Evolution
# ----------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ax = axes[0]
ax.plot(steps, ke_light, color="C0", alpha=0.2)
ax.plot(steps, ke_heavy, color="C1", alpha=0.2)
ax.plot(
    smooth_steps, running_average(ke_light),
    color="C0", label="Light: smoothed group mean",
)
ax.plot(
    smooth_steps, running_average(ke_heavy),
    color="C1", label="Heavy: smoothed group mean",
)
ax.axhline(
    initial_energy / N,
    color="black", linestyle="--", linewidth=1,
    label="Total kinetic energy / N",
)
ax.axvline(
    burn_in, color="grey", linestyle=":", label="Sampling cutoff",
)
ax.set_xlabel("Collision count")
ax.set_ylabel("Mean kinetic energy per particle")
ax.set_title("Redistribution of Kinetic Energy")
ax.legend()
ax.grid(alpha=0.3)

ax = axes[1]
ax.plot(steps, pair_dot_mean, color="purple", alpha=0.2)
ax.plot(
    smooth_steps, running_average(pair_dot_mean),
    color="purple", label="Smoothed pair mean",
)
ax.axhline(0.0, color="black", linestyle="--", linewidth=1)
ax.axvline(burn_in, color="grey", linestyle=":")
ax.set_xlabel("Collision count")
ax.set_ylabel(r"Mean of $\dot{x}_i\cdot\dot{x}_j$, $i\ne j$")
ax.set_title("Scalar Products of Distinct Particle Velocities")
ax.legend()
ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()

# ----------------------------
# Plot 2: Post-cutoff samples
# ----------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Shared bins make the two energy histograms comparable
energy_bins = np.histogram_bin_edges(
    np.concatenate([ke_light_all, ke_heavy_all]), bins=40
)

ax = axes[0]
ax.hist(
    ke_light_all, bins=energy_bins, density=True,
    alpha=0.6, color="C0", label="Light",
)
ax.hist(
    ke_heavy_all, bins=energy_bins, density=True,
    alpha=0.6, color="C1", label="Heavy",
)
ax.set_xlabel("Kinetic energy")
ax.set_ylabel("Probability density")
ax.set_title("Kinetic Energies After Sampling Cutoff")
ax.legend()
ax.grid(alpha=0.3)

ax = axes[1]
ax.hist(
    pair_dot_samples, bins=60, density=True,
    alpha=0.7, color="purple",
)
ax.axvline(0.0, color="black", linestyle="--", linewidth=1)
ax.set_xlabel(r"$\dot{x}_i\cdot\dot{x}_j$, $i\ne j$")
ax.set_ylabel("Probability density")
ax.set_title("Pair Scalar Products After Sampling Cutoff")
ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()
```
