# Sample Answer

## 1. Equations of Motion

The center of mass has body representation $\bar X=l\chi$. Gravity has spatial representation $-mg\chi$ and body representation $-mg\Gamma$. Its moment about the pivot is therefore
$$
T_g=(l\chi)\times(-mg\Gamma)
=mgl\,\Gamma\times\chi.
$$
The pivot reaction acts at $O$, so its moment about $O$ vanishes.

Since $\Pi=\mathbb I\Omega$, the equations in $(R,\Pi)$ are
$$
\boxed{
\dot R=R\widehat{\mathbb I^{-1}\Pi},
\qquad
\dot\Pi=\Pi\times\mathbb I^{-1}\Pi
+mgl(R^T\chi)\times\chi.
}
$$

Differentiating $\Gamma=R^T\chi$ gives
$$
\dot\Gamma=-\widehat\Omega R^T\chi
=\Gamma\times\Omega.
$$
Thus the reduced equations are
$$
\boxed{
\dot\Gamma=\Gamma\times\Omega,
\qquad
\mathbb I\dot\Omega
=(\mathbb I\Omega)\times\Omega+mgl\,\Gamma\times\chi.
}
$$


## 2. Conserved Quantities

Write $c=mgl$. The total mechanical energy is
$$
E=\frac12\Omega^T\mathbb I\Omega+c\,\Gamma^T\chi.
$$
Here $l\Gamma^T\chi$ is the center-of-mass height relative to the pivot. Differentiating,
$$
\begin{aligned}
\dot E
&=\Omega\cdot\left(\Pi\times\Omega+c\,\Gamma\times\chi\right)
+c(\Gamma\times\Omega)\cdot\chi\\
&=0.
\end{aligned}
$$

The vertical component of spatial angular momentum is
$$
J=\Pi^T\Gamma=\chi^T\pi.
$$
Its derivative is
$$
\dot J
=(\Pi\times\Omega+c\,\Gamma\times\chi)\cdot\Gamma
+\Pi\cdot(\Gamma\times\Omega)=0.
$$

The angular momentum component along the body's symmetry axis is
$$
L=\Pi^T\chi=I_3\Omega_3.
$$
Because $I_1=I_2=I_\perp$,
$$
\dot L=(\Pi\times\Omega)\cdot\chi
=(I_\perp-I_\perp)\Omega_1\Omega_2=0.
$$
Finally,
$$
\frac{d}{dt}\|\Gamma\|^2
=2\Gamma\cdot(\Gamma\times\Omega)=0.
$$

Therefore,
$$
\boxed{
E,\quad J,\quad L,\quad \|\Gamma\|^2
\text{ are conserved.}
}
$$

The full spatial angular momentum is generally not constant:
$$
\dot\pi=RT_g
=mgl\,\chi\times(R\chi).
$$
Its magnitude is not generally conserved either:
$$
\frac{d}{dt}\frac12\|\pi\|^2
=mgl\,\Pi\cdot(\Gamma\times\chi).
$$
Only its vertical component is necessarily constant.


## 3. Inclination and Accessible Motion

Let $\varphi$ denote the azimuth of the symmetry axis about the upward vertical, and let $\psi$ denote spin about that axis. For $0<\vartheta<\pi$,
$$
\Omega_3=\dot\psi+\dot\varphi\cos\vartheta,
$$
and
$$
\mathrm{KE}
=\frac12I_\perp
\left(\dot\vartheta^2+\dot\varphi^2\sin^2\vartheta\right)
+\frac12I_3\Omega_3^2.
$$

The conserved momenta give
$$
L=I_3\Omega_3,
\qquad
J=I_\perp\dot\varphi\sin^2\vartheta+L\cos\vartheta.
$$
Consequently,
$$
\dot\varphi
=\frac{J-L\cos\vartheta}{I_\perp\sin^2\vartheta}.
$$
Substituting into the energy yields
$$
\boxed{
\frac12I_\perp\dot\vartheta^2
+U_{\mathrm{eff}}(\vartheta)=E,
}
$$
where
$$
\boxed{
U_{\mathrm{eff}}(\vartheta)
=
\frac{(J-L\cos\vartheta)^2}
{2I_\perp\sin^2\vartheta}
+\frac{L^2}{2I_3}
+mgl\cos\vartheta.
}
$$

The accessible inclinations satisfy $E\ge U_{\mathrm{eff}}$.

- Simple roots of $E=U_{\mathrm{eff}}$ are turning angles, where $\dot\vartheta=0$.
- Constant-inclination motions require both $E=U_{\mathrm{eff}}$ and $U_{\mathrm{eff}}'=0$. Away from the poles, these are regular precessions.
- A strict local minimum gives stable inclination oscillations at fixed $J,L$.
- A nondegenerate local maximum gives an unstable constant-inclination motion. At its energy, nearby motion can approach it asymptotically.

The poles require separate treatment: finite-energy access to $\vartheta=0$ requires $J=L$, while access to $\vartheta=\pi$ requires $J=-L$.


## 4. Upright Steady Spin and Linearization

For
$$
\Gamma=\chi,\qquad \Omega=s\chi,
$$
both cross products in the reduced equations vanish. Hence this is a steady reduced solution, while the top continues to spin.

Write
$$
\Gamma=\chi+\gamma,\qquad
\Omega=s\chi+u.
$$
The unit-length constraint gives $\chi^T\gamma=0$ to first order. Also, $\dot u_3=0$.

Define the planar operator
$$
J_2=\begin{bmatrix}0&-1\\1&0\end{bmatrix},
$$
and let $\gamma_\perp,u_\perp$ contain the first two components. The transverse linearization is
$$
\begin{aligned}
\dot\gamma_\perp
&=J_2(u_\perp-s\gamma_\perp),\\
\dot u_\perp
&=aJ_2u_\perp-bJ_2\gamma_\perp,
\end{aligned}
$$
where
$$
a=\frac{I_3-I_\perp}{I_\perp}s,
\qquad
b=\frac{mgl}{I_\perp}.
$$

Using complex transverse coordinates, the characteristic equation is
$$
\lambda^2+i(s-a)\lambda+(as-b)=0.
$$
Thus
$$
\lambda=\frac{i(a-s)}2
\pm\frac12\sqrt{4b-(a+s)^2},
$$
together with the conjugate pair. Since $a+s=I_3s/I_\perp$,
$$
\boxed{
s_{\mathrm{crit}}
=\frac{2\sqrt{mglI_\perp}}{I_3}.
}
$$

- If $|s|<s_{\mathrm{crit}}$, there is an exponentially growing transverse mode.
- If $|s|>s_{\mathrm{crit}}$, all transverse eigenvalues are purely imaginary, giving oscillatory linearized motion.
- The threshold case requires separate analysis.

Purely imaginary eigenvalues alone do not establish nonlinear stability.


## 5. Numerical Simulations

For the prescribed parameters,
$$
mgl=0.4905\,\mathrm{N\,m},
\qquad
s_{\mathrm{crit}}\approx19.81\,\mathrm{s^{-1}}.
$$
Therefore $s=10\,\mathrm{s^{-1}}$ is below threshold, while $s=30\,\mathrm{s^{-1}}$ is above it.

Let $\vartheta_0=0.05$. The initial data give
$$
\Gamma(0)=
\begin{bmatrix}0\\\sin\vartheta_0\\\cos\vartheta_0\end{bmatrix},
\qquad
\Pi(0)=I_3s\chi,
$$
and hence
$$
E=\frac12I_3s^2+mgl\cos\vartheta_0,
\qquad
J=I_3s\cos\vartheta_0,
\qquad
L=I_3s.
$$

| $s$ ($\mathrm{s^{-1}}$) | $E$ (J) | $J$ ($\mathrm{kg\,m^2\,s^{-1}}$) | $L$ ($\mathrm{kg\,m^2\,s^{-1}}$) |
|---|---:|---:|---:|
| 10 | 0.989887 | 0.0998750 | 0.100000 |
| 30 | 4.989887 | 0.299625 | 0.300000 |

The lower-spin motion departs substantially from the upright configuration. The higher-spin motion remains near it for these small initial perturbations. In both cases, $\Omega_3=s$ and $\Pi_3=I_3s$ remain constant.

The unstable linearization predicts growth only while the perturbations remain small; it does not predict the subsequent large-amplitude motion.


## 6. Effective-Potential Figure

For each run, use its own values of $E,J,L$ to plot $U_{\mathrm{eff}}$. The horizontal line at $E$ intersects the potential at the turning inclinations.

Initially, $\dot\vartheta=0$, so $\vartheta_0=0.05$ is a turning angle. The other turning angle determines the maximum inclination reached by the motion.

For these initial conditions, $J\ne L$ and $J\ne-L$. Therefore $U_{\mathrm{eff}}$ diverges at both poles, confining the inclination to an interval strictly inside $(0,\pi)$.

Instability of the upright spin therefore does not imply unbounded inclination. Energy conservation limits the subsequent motion, and the top nutates between its turning angles.


## 7. Symmetry-Axis Figure

The spatial symmetry-axis direction is
$$
a(t)=R(t)\chi,\qquad \|a(t)\|=1.
$$
Its inclination to the upward vertical satisfies
$$
\chi^Ta(t)=\Gamma(t)^T\chi=\cos\vartheta(t).
$$

The path of $a(t)$ on the unit sphere displays:

- **Nutation:** changes in its polar angle $\vartheta$.
- **Precession:** changes in its azimuth about the vertical.

The higher-spin run remains near the north pole. The lower-spin run explores a wider range of inclinations. Periodic nutation need not produce a closed symmetry-axis path, since the accumulated precession over a nutation period need not be a rational multiple of $2\pi$.


## 8. Animation and Physical Interpretation

The spatial panel shows the actual motion of the body and its angular momentum. Gravity changes $\pi$ through
$$
\dot\pi=mgl\,\chi\times(R\chi),
$$
but this moment has no vertical component, so $\chi^T\pi=J$ remains constant.

In the body panel, the top is stationary while the upward vertical direction $\Gamma$ moves. The symmetry-axis momentum $L=\Pi^T\chi$ remains constant because gravity has no moment along that axis and the transverse inertias are equal.

A constant inclination does not imply a stationary orientation: the top can precess about the vertical while simultaneously spinning about its symmetry axis.


## 9. Numerical Verification

Monitor
$$
E(t)-E(0),\qquad
J(t)-J(0),\qquad
L(t)-L(0),
$$
together with
$$
\|\Gamma\|^2-1,\qquad
\|\Gamma-R^T\chi\|,\qquad
\|R^TR-I\|_F,\qquad
|\det R-1|.
$$

Repeat the calculation with tighter tolerances or a smaller time step. The inclination extrema, precession, and conserved quantities should converge.

If $\Gamma=R^T\chi$ is calculated directly, its consistency is automatic. If rotation matrices are reconstructed from normalized quaternions, orthogonality is likewise enforced by the representation; these checks do not replace monitoring energy and momentum conservation.

## Animation

```python
import numpy as np
import plotly.graph_objects as go

from scipy.integrate import solve_ivp
from scipy.optimize import brentq
from scipy.spatial.transform import Rotation
from plotly.subplots import make_subplots

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------
m, l, g = 0.5, 0.1, 9.81
Ip, I3 = 0.02, 0.01
II = np.diag([Ip, Ip, I3])
invII = np.diag(1.0 / np.diag(II))
chi = np.array([0.0, 0.0, 1.0])
theta0 = 0.05
Tmax = 10.0
times = np.linspace(0.0, Tmax, 1501)
c = m * g * l

scrit = 2 * np.sqrt(c * Ip) / I3
print(f"Critical spin: {scrit:.6f} rad/s")


def quaternion_product(q, p):
    """Hamilton product, scalar-last convention [x, y, z, w]."""
    v, w = q[:3], q[3]
    z, a = p[:3], p[3]
    return np.r_[w*z + a*v + np.cross(v, z), w*a - v @ z]


def simulate_top(s, rtol=1e-9, atol=1e-11):
    R0 = Rotation.from_rotvec(theta0 * np.array([1., 0., 0.]))
    q0 = R0.as_quat()
    pi0 = R0.as_matrix() @ (II @ (s * chi))

    def rhs(t, state):
        q, pi = state[:4], state[4:]
        R = Rotation.from_quat(q).as_matrix()
        omega = R @ invII @ R.T @ pi
        torque = c * np.cross(chi, R @ chi)

        # Spatial angular velocity: qdot = 1/2 [omega, 0] ⊗ q.
        qdot = 0.5 * quaternion_product(np.r_[omega, 0.0], q)
        return np.r_[qdot, torque]

    sol = solve_ivp(
        rhs, (0.0, Tmax), np.r_[q0, pi0],
        t_eval=times, method="DOP853",
        rtol=rtol, atol=atol,
    )
    if not sol.success:
        raise RuntimeError(sol.message)

    q = sol.y[:4].T
    pi = sol.y[4:].T
    R = Rotation.from_quat(q).as_matrix()
    Pi = np.einsum("nji,nj->ni", R, pi)
    Omega = Pi @ invII
    omega = np.einsum("nij,nj->ni", R, Omega)
    Gamma = R[:, 2, :]       # R^T chi
    axis = R[:, :, 2]        # R chi
    torque = c * np.cross(chi, axis)

    energy = 0.5 * np.sum(Pi * Omega, axis=1) + c * Gamma[:, 2]
    J = np.sum(Pi * Gamma, axis=1)
    L = Pi[:, 2]
    theta = np.arccos(np.clip(Gamma[:, 2], -1, 1))

    return dict(
        s=s, t=sol.t, R=R, pi=pi, Pi=Pi,
        Omega=Omega, omega=omega, Gamma=Gamma,
        axis=axis, torque=torque, E=energy, J=J, L=L,
        theta=theta, q=q,
    )


runs = [simulate_top(10.0), simulate_top(30.0)]

# ------------------------------------------------------------
# Component and inclination plots
# ------------------------------------------------------------
colors = ["#154C79", "#176B45", "#D1495B"]

for run in runs:
    fig = make_subplots(
        rows=3, cols=1, shared_xaxes=True,
        subplot_titles=[
            "Body angular velocity",
            "Body angular momentum",
            "Inclination",
        ],
        vertical_spacing=0.09,
    )

    for j in range(3):
        fig.add_trace(go.Scatter(
            x=run["t"], y=run["Omega"][:, j],
            name=f"Ω{j+1}", line=dict(color=colors[j]),
        ), row=1, col=1)

        fig.add_trace(go.Scatter(
            x=run["t"], y=run["Pi"][:, j],
            name=f"Π{j+1}",
            line=dict(color=colors[j], dash="dash"),
        ), row=2, col=1)

    fig.add_trace(go.Scatter(
        x=run["t"], y=np.rad2deg(run["theta"]),
        name="Inclination", line=dict(color="#7C3AED"),
    ), row=3, col=1)

    fig.update_yaxes(title_text="rad/s", row=1, col=1)
    fig.update_yaxes(title_text="kg m²/s", row=2, col=1)
    fig.update_yaxes(title_text="degrees", row=3, col=1)
    fig.update_xaxes(title_text="Time (s)", row=3, col=1)
    fig.update_layout(
        title=f"Heavy top: initial spin {run['s']:g} rad/s",
        height=800, template="plotly_white",
    )
    fig.show()

# ------------------------------------------------------------
# Effective potentials and turning angles
# ------------------------------------------------------------
def effective_potential(theta, J, L):
    return (
        (J - L * np.cos(theta))**2
        / (2 * Ip * np.sin(theta)**2)
        + L**2 / (2 * I3)
        + c * np.cos(theta)
    )


fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=[f"s = {run['s']:g} rad/s" for run in runs],
)

for column, run in enumerate(runs, start=1):
    E0, J0, L0 = run["E"][0], run["J"][0], run["L"][0]

    def residual(theta):
        return effective_potential(theta, J0, L0) - E0

    scan = np.linspace(1e-5, np.pi - 1e-5, 20001)
    values = residual(scan)
    crossings = np.where(values[:-1] * values[1:] < 0)[0]
    roots = np.array([
        brentq(residual, scan[k], scan[k + 1])
        for k in crossings
    ])

    if len(roots) != 2:
        raise RuntimeError("Expected two simple turning angles for these ICs.")

    run["turning_angles"] = roots
    lower, upper = roots
    span = upper - lower
    theta_grid = np.linspace(
        max(1e-5, lower - 0.2 * span),
        min(np.pi - 1e-5, upper + 0.2 * span),
        1000,
    )
    U = effective_potential(theta_grid, J0, L0)

    fig.add_trace(go.Scatter(
        x=np.rad2deg(theta_grid), y=U,
        name=f"Ueff, s={run['s']:g}",
        line=dict(color="#154C79"),
    ), row=1, col=column)

    fig.add_trace(go.Scatter(
        x=np.rad2deg([theta_grid[0], theta_grid[-1]]),
        y=[E0, E0], name=f"E, s={run['s']:g}",
        line=dict(color="#D1495B", dash="dash"),
    ), row=1, col=column)

    fig.add_trace(go.Scatter(
        x=np.rad2deg(roots), y=[E0, E0],
        mode="markers", name="Turning angles",
        marker=dict(size=9, color="#176B45"),
        showlegend=(column == 1),
    ), row=1, col=column)

    fig.update_xaxes(title_text="Inclination (degrees)", row=1, col=column)
    fig.update_yaxes(title_text="Energy (J)", row=1, col=column)

    print(
        f"s={run['s']:g}: predicted turning angles "
        f"{np.rad2deg(roots)} degrees; simulated range "
        f"{np.rad2deg([run['theta'].min(), run['theta'].max()])}"
    )

fig.update_layout(
    title="Effective potentials and accessible inclinations",
    template="plotly_white", height=450, width=1050,
)
fig.show()

# ------------------------------------------------------------
# Symmetry-axis paths on a unit sphere
# ------------------------------------------------------------
polar, azimuth = np.meshgrid(
    np.linspace(0, np.pi, 60),
    np.linspace(0, 2*np.pi, 100),
    indexing="ij",
)

fig = go.Figure(go.Surface(
    x=np.sin(polar)*np.cos(azimuth),
    y=np.sin(polar)*np.sin(azimuth),
    z=np.cos(polar),
    surfacecolor=np.zeros_like(polar),
    colorscale=[[0, "#DCE3EA"], [1, "#DCE3EA"]],
    showscale=False, opacity=0.2, hoverinfo="skip",
))

for run, color in zip(runs, ["#154C79", "#D1495B"]):
    path = run["axis"]
    fig.add_trace(go.Scatter3d(
        x=path[:, 0], y=path[:, 1], z=path[:, 2],
        mode="lines", name=f"s={run['s']:g} rad/s",
        line=dict(color=color, width=5),
    ))

    # Direction arrows from adot = omega x a.
    indices = np.linspace(0, len(path)-1, 7, dtype=int)[1:-1]
    velocity = np.cross(run["omega"][indices], path[indices])
    velocity /= np.linalg.norm(velocity, axis=1, keepdims=True)

    fig.add_trace(go.Cone(
        x=path[indices, 0], y=path[indices, 1], z=path[indices, 2],
        u=velocity[:, 0], v=velocity[:, 1], w=velocity[:, 2],
        sizemode="absolute", sizeref=0.07,
        colorscale=[[0, color], [1, color]],
        showscale=False, showlegend=False,
    ))

fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, 0], z=[0, 1.2],
    mode="lines+text", text=["", "Upward vertical"],
    line=dict(color="black", width=4), name="Vertical",
))

fig.update_layout(
    title="Spatial motion of the symmetry axis",
    height=700,
    scene=dict(
        aspectmode="cube",
        xaxis=dict(range=[-1.2, 1.2], title="e₁"),
        yaxis=dict(range=[-1.2, 1.2], title="e₂"),
        zaxis=dict(range=[-1.2, 1.2], title="e₃"),
    ),
    sliders=[dict(
        currentvalue=dict(prefix="Sphere opacity: "),
        active=2,
        steps=[
            dict(
                method="restyle",
                args=[{"opacity": value}, [0]],
                label=f"{value:.1f}",
            )
            for value in np.linspace(0, 1, 11)
        ],
    )],
)
fig.show()

# ------------------------------------------------------------
# Conservation and convergence diagnostics
# ------------------------------------------------------------
for run in runs:
    print(f"\nInitial spin: {run['s']:g} rad/s")

    for key in ["E", "J", "L"]:
        error = np.max(np.abs(run[key] - run[key][0]))
        print(f"  Maximum absolute {key} drift: {error:.3e}")

    RTR = np.einsum("nji,njk->nik", run["R"], run["R"])
    print(
        "  Maximum orthogonality error:",
        f"{np.max(np.linalg.norm(RTR-np.eye(3), axis=(1, 2))):.3e}"
    )
    print(
        "  Maximum raw quaternion norm error:",
        f"{np.max(np.abs(np.linalg.norm(run['q'], axis=1)-1)):.3e}"
    )

    refined = simulate_top(run["s"], rtol=1e-11, atol=1e-13)
    print(
        "  Maximum inclination difference after refinement:",
        f"{np.max(np.abs(run['theta']-refined['theta'])):.3e} rad"
    )
```
