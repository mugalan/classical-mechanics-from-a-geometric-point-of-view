# Assignment on the Motion of a Axi-Symmetric Top

axisymmetric-top.png

Consider an axisymmetric rigid body with a fixed,
frictionless pivot $O$. Let $\mathbf e$ be an inertial
frame with $\mathbf e_3$ pointing vertically upwards,
and let $\mathbf b=\mathbf eR$ be a body frame with
origin at $O$. Define
$$
\chi=\begin{bmatrix}0\\0\\1\end{bmatrix}.
$$
The center of mass has body coordinates $l\chi$,
and the inertia tensor about the pivot is
$$
\mathbb{I}=\operatorname{diag}(I_\perp,I_\perp,I_3).
$$
Gravity is the only interaction producing a moment
about the pivot. Define
$$
\Gamma=R^T\chi,\qquad \Pi=\mathbb{I}\Omega.
$$

Answer the following.

* Derive the gravitational force moment in the
body frame and obtain the reduced equations
$$
\dot{\Gamma}=\Gamma\times\Omega,\qquad
\mathbb{I}\dot{\Omega}
=(\mathbb{I}\Omega)\times\Omega
+mgl\,\Gamma\times\chi.
$$

* Show that the following quantities are conserved:
$$
E=\frac12\Omega^T\mathbb{I}\Omega+mgl\,\Gamma^T\chi,
\qquad
\Pi^T\Gamma,\qquad
\Pi^T\chi,\qquad
\|\Gamma\|^2.
$$
Explain the physical meaning of the first three.
Does the full spatial angular momentum $R\Pi$
remain constant?

* Show that
$$
\Gamma=\chi,\qquad \Omega=s\chi
$$
describes an upright steady spin.
Linearize the reduced equations about this motion,
using perturbations consistent with $\|\Gamma\|=1$.
Determine the spin threshold separating exponentially
growing transverse perturbations from oscillatory
ones. You need not classify the threshold case.

* Simulate the motion with
$$
m=0.5\,\mathrm{kg},\qquad
l=0.1\,\mathrm{m},\qquad
g=9.81\,\mathrm{m\,s^{-2}},
$$
$$
I_\perp=0.02\,\mathrm{kg\,m^2},\qquad
I_3=0.01\,\mathrm{kg\,m^2},
$$
and initial conditions
$$
R(0)=R_1(0.05),\qquad
\Gamma(0)=R(0)^T\chi,\qquad
\Omega(0)=s\chi.
$$
Compare $s=10\,\mathrm{s^{-1}}$ and
$s=30\,\mathrm{s^{-1}}$ over a time interval
of at least $10\,\mathrm{s}$.

* Plot the angular velocity components and the
inclination angle
$$
\vartheta(t)=\cos^{-1}\bigl(\Gamma(t)^T\chi\bigr).
$$
Relate the observed motion to the threshold found
above.

* Integrate $\dot R=R\widehat{\Omega}$ to
reconstruct the orientation, and plot the path of
the symmetry-axis direction $R(t)\chi$ on the unit
sphere. Describe its precession and nutation.

* Monitor the conserved quantities from part (ii),
the consistency condition $\Gamma=R^T\chi$, and
the orthogonality of $R$. Refine the numerical
tolerances or time step to verify the results.
