# Assignment on Free Rigid Body Motion

Consider a uniform solid rectangular block of mass
$M=10\,\mathrm{kg}$, length $l=2\,\mathrm{m}$,
width $w=1\,\mathrm{m}$, and height $h=4\,\mathrm{m}$,
moving with no external forces or force moments.

Let $\mathbf e$ be an inertial frame, and let
$\mathbf b=\mathbf eR$ be a body-fixed frame with
origin at the center of mass and axes aligned with
the length, width, and height of the block, respectively.
These are principal axes, and the inertia tensor is
$$
\mathbb{I}
=\frac{M}{12}\operatorname{diag}
\left(w^2+h^2,\ l^2+h^2,\ l^2+w^2\right)
=\operatorname{diag}
\left(\frac{85}{6},\frac{50}{3},\frac{25}{6}\right)
\,\mathrm{kg\,m^2}.
$$
Let $\Pi=\mathbb{I}\Omega$ be the body representation
of angular momentum about the center of mass.
Answer the following:

* Write down the rotational equations of motion using $(R,\Pi)$ as the variables. Show that
  $$
  \pi=R\Pi,\qquad
  E=\frac12\Pi^T\mathbb I^{-1}\Pi,\qquad
  h^2=\Pi^T\Pi
  $$
  are conserved.

* Identify all nonzero steady rotations about the principal axes. Explain why these are called **relative equilibria**, although the orientation $R$ continues to change.

* Let $\mathbb I_{\max}>\mathbb I_{\mathrm{int}}>\mathbb I_{\min}>0$ denote the ordered principal moments. Show that the energy ellipsoid and angular momentum sphere intersect if and only if
  $$
  \frac{h^2}{2\mathbb I_{\max}}
  \le E\le
  \frac{h^2}{2\mathbb I_{\min}}.
  $$
  Explain the intersections at the two bounds and identify the corresponding relative equilibria. Deduce the equivalent condition on $h$ when $E>0$ is fixed.

* Define
  $$
  E_{\mathrm{crit}}=\frac{h^2}{2\mathbb I_{\mathrm{int}}}.
  $$
  Describe the intersection geometry and the resulting motion of $\Pi$ in the regimes
  $$
  E=E_{\min},\quad
  E_{\min}<E<E_{\mathrm{crit}},\quad
  E=E_{\mathrm{crit}},\quad
  E_{\mathrm{crit}}<E<E_{\max},\quad
  E=E_{\max}.
  $$
  Identify which relative equilibria the regular curves surround. At the critical energy, show that the intersection consists of two great circles meeting at the intermediate-axis equilibria. Explain why the four connecting branches carry nonperiodic solutions, although the complete geometric intersection consists of closed circles.

  $$

* Linearize the body angular momentum equations about each nonzero steady rotation. Determine which rotations have oscillatory transverse perturbations and which have exponentially growing transverse perturbations. Relate these results to the intersection geometry.

* Simulate the block's rotational motion with $R(0)=I_{3\times3}$ and each of the initial conditions
  $$

  \qquad
